```python
# if, elif
score = float(input())
if 0 >= score < 0.3:
    status = 'bad'
elif 0.3 >= score < 0.5:
    status = 'not_so_good'
elif 0.5 >= score < 0.7:
    status = 'good'
elif 0.7 >= score <= 1:
    status = 'excellent'
else:
    status = 'wrong'

print(status)
```

```python
## делится на 3 Fizz, делится на 5 Buzz, на 3 и 5 = FizzBuzz, иначе само число

number = int(input())

if (number % 3 == 0 and number % 5 == 0):
    print('FizzBuzz')
elif (number % 3 == 0 and number % 5 != 0):
    print('Fizz')
elif (number % 3 != 0 and number % 5 == 0):
    print('Buzz')
else:
    print(number)
```

```python
# Вывести количество совпадающих цифр

a, b, c = map(int, input().split())

if a == b and b == c:
    print(3)
elif (a == b and b != c) or (a == c and c != b) or (b == c and c!= a):
    print(2)
else:
    print(0)
```

```python
'''На вход программе поступает предложение. Ваша задача — по последнему символу определить тип предложения следующим образом:
    ➖ если предложение заканчивается точкой, то оно считается законченным;
    ➖ если предложение заканчивается запятой, то оно считается незаконченным;
    ➖ если предложение заканчивается знаком ?, то оно считается вопросительным;
    ➖ если предложение заканчивается знаком !, то оно считается восклицательным;
    ➖ в остальных случаях считаем предложение неопределенным.'''

sentence = input()
last_char = list(sentence)[-1]
sentence_type = None

if last_char == '.':
    sentence_type = 'Законченное'
elif last_char == ',':
    sentence_type = 'Незаконченное'
elif last_char == '!':
    sentence_type = 'Восклицательное'
elif last_char == '?':
    sentence_type = 'Вопросительное'
else:
    sentence_type = 'Неопределенное'

print(sentence_type)
```
