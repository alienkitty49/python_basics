```python
# Проверяем накопления
# break
earnings = list(map(int, input().split()))
goal = int(input())

i = 0
current_sum = 0
is_reached = False

while i < len(earnings):
    current_sum += earnings[i]
    if current_sum >= goal:
        is_reached = True
        break
    i += 1

print(current_sum)
if is_reached:
    print("Цель достигнута")
else:
    print("Цель не достигнута")
```

```python
'''
Вводятся два натуральных числа a и b в отдельных строках. Гарантируется, что a<b.
Вывести все числа на интервале от а до b включительно согласно условиям в пунктах 1 и 2
При этом нужно:
    - пропускать (не выводить) числа, которые делятся на 2 или на 3;
    - если встречаете число, кратное 777, необходимо принудительно закончить цикл, само это число не выводить
'''
a = int(input())
b = int(input())

i = 0
while (a + i) <= b:
    if (a + i) % 2 != 0 and (a + i) % 3 != 0 and (a + i) % 777 != 0:
        print(a + i)
    if (a + i) % 777 == 0:
        break
    i += 1

```

```python
'''Представь, что у тебя есть поток чисел, и ты хочешь вывести только те, которые больше 5 и не равны 7. Для значений ≤ 5 или равных 7 используй continue. Цикл должен обработать числа от 1 до 12'''

i = 0

while i <= 12:
    i += 1
    if i <= 5 or i == 7:
        continue
    print(i)
```

```python
'''На вход программе поступает строка, ваша задача - обойти в обратном порядке все символы данной строки и вывести пару значений: текущий символ введенной строки и его индекс.
'''

s = input()
i = 1
chars = len(s)# количество символов в строке

while i <= len(s): # идем по строке пока она не закончится
    print(f'{s[-i]} {chars-1}') # последний символ и его индекс (убаляем длину не единицу, т.к. индексу в нуля
    i += 1
    chars -= 1
```

```python
'''Пользователь вводит целые числа по одному в строке, последовательность оканчивается числом 0. Все, что вводится после 0, не относится к последовательности, ваша программа не должна их считывать. Напишите программу, которая выводит сумму всех членов данной последовательности.'''

nums = list(map(int, input().split()))
i = 0
summa = 0
while i < len(nums):
    i += 1
    if nums[i-1] != 0:
        summa += nums[i-1]
    else:
        break
print(summa)
```

```python
'''На каждой отдельной строчке пользователь вводит друг за другом пароли в виде строки символов. Валидными паролями будем считать строки, у которых длина варьируется от 5 до 9 символов включительно. Как только вы встретите первый невалидный пароль, ваша программа должна закончить считывать пароли и вывести последний введенный валидный пароль.

Гарантируется, что первый пароль всегда валидный
'''


s = list(map(str, input().split()))

i = 0
p = None

while i < len(s):
    p = s[i]
    if len(p) < 5 or len(p) > 9:
        print(s[i-1])
        break
    i += 1
```

```python
'''Напишите программу, моделирующую изменение здоровья персонажа в игре. Изначально у героя 100 единиц здоровья. В ходе игры противник наносит удары, каждый из которых уменьшает здоровье персонажа на определённое количество очков. Значения урона поступают на вход программе в отдельных строках.

Удары продолжаются до тех пор, пока здоровье персонажа не станет равным нулю или меньше нуля.'''

fights = list(map(int, input().split()))
h = 100
i = 0

print(f'Уровень здоровья: {h}')

while h >= 0:
    h = h - fights[i]
    if h >=0:
        print(f'Уровень здоровья: {h}')
    else:
        print(f'Игра окончена')
    i += 1

print(f'Количество ударов, которые ваш персонаж героически выдержал = {i-1}')
print(f'Удар № {i} был критическим')
```

```python
# проверка на палиндром

s = input()

left, right = 0, (len(s)-1) #первый и последний индекс символа в строке
is_palindrome = 'Палиндром'

while left < right:
    if s[left] != s[right]:
        is_palindrome = 'Не палиндром'
        break
    else:
        left += 1
        right -=1

print(is_palindrome)
```

```python
# Количество цифр, кратных 2, 3, 5

n = input()

i = 0
kratno_2 = 0
kratno_3 = 0
kratno_5 = 0


while i < len(n):
    if int(n[i]) % 2 == 0:
        kratno_2 += 1
    elif int(n[i]) % 3 == 0:
        kratno_3 += 1
    elif int(n[i]) % 5 == 0:
        kratno_5 += 1
    i += 1
print(f'Всего цифр: {i}')
print(f'Кратно 2: {kratno_2} \nКратно 3: {kratno_3} \nКратно 5: {kratno_5}')
```
